```python
import sys
sys.path.insert(0, "../..")

import numpy as np
import matplotlib.pyplot as plt
from tests.test_signal import make_pulse
```

```python
fs = 96000
duration = 2.0
onset = 1.6
tau = 0.15e-3

t, force = make_pulse(tau, fs, duration, onset)
```

```python
plt.figure(figsize=(10, 4))
window = (t > onset - 0.1e-3) & (t < onset + tau + 0.1e-3)

plt.plot(t[window], force[window], marker='.')
plt.xlabel("time (s)")
plt.ylabel("force")
plt.title("Synthetic pulse, tau = 0.15 ms")
plt.show
```

```python
mask = (t >= onset) & (t <= (onset + tau))
count = np.sum(mask)
predicted = tau * fs
print("tau =", tau, ", predicted N =", predicted, ", measured N =", count)
```
